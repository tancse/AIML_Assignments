```python
import numpy as np

data = np.array([[22.5, 19.0, 31.2, 28.7, 25.1],
                 [17.3, 22.8, 30.5, 26.4, 21.9],
                 [33.1, 29.6, 18.4, 24.0, 27.8],
                 [20.2, 23.5, 31.9, 28.1, 22.6]])
```

#Task 1

```python
print(np.shape(data))
mean_per_station = data.mean(axis=1)
for i, mean in enumerate(mean_per_station):
    print(f"Station {i+1} mean temperature: {mean:.2f}")
```

#Task 2


```python
mask = data > 28.0
print(f"Boolean mask:\n{mask}")

above_28 = data[mask]
print(f"\n1D array of readings above 28.0:\n{above_28}")
```

#Task 3

```python
normalized = (data - data.min()) / (data.max() - data.min())
print(f"Normalized array (rounded to 2 decimal places):\n{np.round(normalized, 2)}")
```
